```python
import sympy as sp
from isb import ISB

# variables simbolicas
t, s = sp.symbols("t s")

# funciones
x_t = sp.E**(-2*t)*sp.cos(2*t)

sp.plot(x_t, (t, 0, 5), title=r"Oscilación Amortiguada: $x(t) = e^{-2t} \cos(2t)$");
```

```python
# pasar a Laplace
X_s = sp.laplace_transform(x_t, t, s, noconds=True)
X_s
```

```python
# graficar 

isb = ISB()

isb.plot_pz(X_s, t_range=(0, 10));
```

```python
isb.pzmap(X_s, t_range=(0, 6));
```
