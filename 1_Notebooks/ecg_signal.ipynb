```python
import numpy as np
import sympy as sp

# Define time variable and parameters
t = sp.symbols('t')

# Define parameters for P, Q, R, S, T waves (amplitude, center time, width)
params = {
    'a_P': 0.15, 't_P': 0.2, 's_P': 0.025,
    'a_Q': -0.05, 't_Q': 0.35, 's_Q': 0.01,
    'a_R': 1.0, 't_R': 0.4, 's_R': 0.015,
    'a_S': -0.2, 't_S': 0.45, 's_S': 0.01,
    'a_T': 0.3, 't_T': 0.6, 's_T': 0.04,
}

# Construct the parametric symbolic expression
ecg_expr = sum(
    p['a'] * sp.exp(-((t - p['t']) ** 2) / (2 * p['s'] ** 2))
    for p in [
        {'a': params['a_P'], 't': params['t_P'], 's': params['s_P']},
        {'a': params['a_Q'], 't': params['t_Q'], 's': params['s_Q']},
        {'a': params['a_R'], 't': params['t_R'], 's': params['s_R']},
        {'a': params['a_S'], 't': params['t_S'], 's': params['s_S']},
        {'a': params['a_T'], 't': params['t_T'], 's': params['s_T']},
    ]
)

display(ecg_expr)
```

```python
sp.plot(ecg_expr, (t,0,2.5))
```
