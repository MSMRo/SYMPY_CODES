```python
import sympy as sp
from sympy.physics.control.lti import TransferFunction

s = sp.symbols('s')

# Define a second-order system
num = s+10
den = s**2 + 3*s + 10
G = TransferFunction(num, den, s)

G
```

```python
poles = G.poles()
zeros = G.zeros()

poles, zeros
```

```python
from isb_lib import ISB


ISB.plot_pz(G);
```

## Ejemplo sistema 2do orden

```python
# Define symbols and a symbolic expression
s = sp.symbols('s')
expr = (s+10) / (s**2 + 3*s + 10)

expr
```

```python
ISB.plot_pz(expr);
```

## Ejemplo sistemas

```python
from sympy.physics.control.lti import Series, Parallel, Feedback

G1 = TransferFunction(1, s + 1, s)
G2 = TransferFunction(5, s + 5, s)

# Series connection: G1 * G2
sys_series = Series(G1, G2)
display(sys_series.rewrite(TransferFunction))

# Parallel connection: G1 + G2
sys_parallel = Parallel(G1, G2)
display(sys_parallel.rewrite(TransferFunction))

# Closed-loop feedback system with negative feedback: G1 / (1 + G1 * G2)
sys_feedback = Feedback(G1, G2)
display(sys_feedback.rewrite(TransferFunction))
```
