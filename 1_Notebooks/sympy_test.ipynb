```python
import sympy as sp
```

```python
t=sp.symbols("t")

x = sp.E**(-0.1*t)*sp.cos(2*t)

sp.plot(x, (t,0,10))
```

```python
from isb_lib import ISB

X_s, x_t, polos, ceros = ISB.plot_pz(x, t_range=(0, 10))

print("Polos:", polos)
print("Ceros:", ceros)
```
