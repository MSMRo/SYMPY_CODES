## Gráficas de funciones

```python
import sympy as sp

# Inicializa la visualización matemática bonita en consolas o Jupyter Notebooks
sp.init_printing()

# Declaración de variables simbólicas
x, y = sp.symbols('x y')
```

```python
# Definir una función: f(x) = x^3 - 5x^2 + 4
f = x**3 - 5*x**2 + 4

# Graficar f(x) = x^3 - 5x^2 + 4 en el rango x ∈ [-3, 3]
sp.plot(f, (x, -3, 3), title="Gráfico de $f(x) = x^3 - 5x^2 + 4$")
```

## DErivadas e integrales

```python

# Primera derivada: f'(x)
df_dx = sp.diff(f, x)
print("Primera derivada:", df_dx)  # 3*x**2 - 10*x

# Segunda derivada: f''(x)
d2f_dx2 = sp.diff(f, x, 2)
print("Segunda derivada:", d2f_dx2)  # 6*x - 10
```

```python
# Derivadas parciales (para varias variables): g(x, y) = x^2 * y^3
g = x**2 * y**3
dg_dx = sp.diff(g, x)  # respecto a x: 2*x*y^3
dg_dy = sp.diff(g, y)  # respecto a y: 3*x^2*y^2
```

```python
f = x**2 + sp.sin(x)

# Integral Indefinida: ∫ (x² + sin(x)) dx
# Nota: SymPy no añade la constante C explícitamente
integral_indef = sp.integrate(f, x)
print("Integral indefinida:", integral_indef)  # x**3/3 - cos(x)

# Integral Definida: ∫_a^b f(x) dx
# Pasa una tupla con la variable y sus límites: (variable, límite_inferior, límite_superior)
integral_def = sp.integrate(f, (x, 0, sp.pi))
print("Integral definida (exacta):", integral_def)  # pi**3/3 + 2

# Para obtener el valor numérico decimal de la solución:
print("Valor aproximado:", integral_def.evalf())
```
