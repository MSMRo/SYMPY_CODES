# Pruebas de señales y sistemas


## Ejemplo 1

$x = e^{-2t}*cos(2t)$

```python
import sympy as sp
from isb import ISB
```

```python
s,t = sp.symbols("s t")
```

```python
x = sp.E**(-0.6*t)*sp.cos(10*t)
x
```

```python
X_s = sp.laplace_transform(x, t,s, noconds=True)
X_s
```

```python
#sp.plot(x, (t,0,10))
```

```python
ISB.plot_pz(X_s)
```

## Ejemplo 2

$x = e^{-2t}$

```python
x = sp.E**(-10*t)
x
```

```python
X_s = sp.laplace_transform(x,t,s,noconds=True)
X_s
```

```python
ISB.plot_pz(X_s,  t_range=(0, 20))
```

## Ejemplo 3

$x = sin(2t)$

```python
x = sp.sin(2*t)

X_s = sp.laplace_transform(x,t,s,noconds=True)

ISB.plot_pz(X_s)
```
